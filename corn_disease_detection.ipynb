```python
from PIL import Image
import os

dataset_path = "images_data"

rgb_count = 0
grayscale_count = 0

for folder in os.listdir(dataset_path):
    folder_path = os.path.join(dataset_path, folder)

    if os.path.isdir(folder_path):

        for image_name in os.listdir(folder_path):

            image_path = os.path.join(folder_path, image_name)

            try:
                img = Image.open(image_path)

                if img.mode == "RGB":
                    rgb_count += 1

                elif img.mode == "L":
                    grayscale_count += 1

            except:
                pass

print("RGB Images:", rgb_count)
print("Grayscale Images:", grayscale_count)
```

```python
import os
import hashlib

dataset_path = "images_data"

hashes = {}
duplicates = []

for folder in os.listdir(dataset_path):

    folder_path = os.path.join(dataset_path, folder)

    if os.path.isdir(folder_path):

        for image_name in os.listdir(folder_path):

            image_path = os.path.join(folder_path, image_name)

            try:
                with open(image_path, "rb") as file:
                    file_hash = hashlib.md5(file.read()).hexdigest()

                if file_hash in hashes:
                    duplicates.append((hashes[file_hash], image_path))
                else:
                    hashes[file_hash] = image_path

            except Exception as e:
                print(f"Error reading {image_path}: {e}")

print(f"\nTotal Duplicate Images: {len(duplicates)}")

if duplicates:
    print("\nDuplicate Image Pairs:\n")
    for original, duplicate in duplicates:
        print("Original :", original)
        print("Duplicate:", duplicate)
        print("-" * 60)
else:
    print("No duplicate images found.")
```

```python
import tensorflow as tf

dataset = tf.keras.utils.image_dataset_from_directory(
    "images_data",
    image_size=(224, 224),
    batch_size=32
)
```

```python
print(dataset.class_names)
```

```python
for images, labels in dataset.take(1):
    print(images.shape)
    print(labels.shape)
```

```python
dataset = dataset.map(lambda x, y: (x / 255.0, y))
```

```python
for images, labels in dataset.take(1):
    print(images[0].numpy().min())
    print(images[0].numpy().max())
```

```python
dataset_size = len(dataset)

train_size = int(0.7 * dataset_size)
val_size = int(0.15 * dataset_size)
test_size = dataset_size - train_size - val_size

train_dataset = dataset.take(train_size)

remaining = dataset.skip(train_size)

val_dataset = remaining.take(val_size)
test_dataset = remaining.skip(val_size)

print("Train Batches:", len(train_dataset))
print("Validation Batches:", len(val_dataset))
print("Test Batches:", len(test_dataset))
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential([

    Input(shape=(224,224,3)),

    Conv2D(32, (3,3), activation="relu"),
    MaxPooling2D(),

    Conv2D(64, (3,3), activation="relu"),
    MaxPooling2D(),

    Flatten(),

    Dense(128, activation="relu"),

    Dense(4, activation="softmax")

])

model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=10
)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history["accuracy"])
plt.plot(history.history["val_accuracy"])

plt.title("Model Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend(["Train", "Validation"])

plt.show()
```

```python
plt.plot(history.history["loss"])
plt.plot(history.history["val_loss"])

plt.title("Model Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend(["Train", "Validation"])

plt.show()
```

```python
test_loss, test_accuracy = model.evaluate(test_dataset)

print("Test Accuracy:", test_accuracy)
print("Test Loss:", test_loss)
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix, classification_report
```

```python
actual_labels = []
predicted_labels = []

for images, labels in test_dataset:

    predictions = model.predict(images)

    predicted = np.argmax(predictions, axis=1)

    actual_labels.extend(labels.numpy())
    predicted_labels.extend(predicted)
```

```python
cm = confusion_matrix(actual_labels, predicted_labels)

print(cm)
```

```python
print(classification_report(
    actual_labels,
    predicted_labels,
    target_names=class_names
))
```

```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing import image

# Classes
class_names = [
    "Blight",
    "Common_Rust",
    "Gray_Leaf_Spot",
    "Healthy"
]

# User se class ka naam lo
selected_class = input("Enter Class Name (Blight, Common_Rust, Gray_Leaf_Spot, Healthy): ")

# Check class
if selected_class not in class_names:
    print("Invalid Class Name!")

else:

    # Folder path
    folder_path = os.path.join("images_data", selected_class)

    # Random image
    image_name = random.choice(os.listdir(folder_path))

    # Complete path
    image_path = os.path.join(folder_path, image_name)

    # Load image
    img = image.load_img(image_path, target_size=(224,224))

    # Image to array
    img_array = image.img_to_array(img)

    # Batch dimension
    img_array = np.expand_dims(img_array, axis=0)

    # Normalize
    img_array = img_array / 255.0

    # Prediction
    prediction = model.predict(img_array, verbose=0)

    predicted_index = np.argmax(prediction)

    predicted_class = class_names[predicted_index]

    confidence = np.max(prediction) * 100

    # Show image
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"Prediction: {predicted_class}")
    plt.show()

    print("Selected Class :", selected_class)
    print("Image Name     :", image_name)
    print("Predicted Class:", predicted_class)
    print(f"Confidence     : {confidence:.2f}%")

    if selected_class == predicted_class:
        print("✅ Prediction Correct")
    else:
        print("❌ Prediction Incorrect")
```
